```python

import numpy as np
def analyze_row_patterns(sparse_matrix):
    """
    Analyze patterns in each row of a sparse matrix.
    Returns dictionary with various row statistics.
    """
    # Convert to CSR format for efficient row operations
    # stores non xero by row so each row is a user and don't store zero's 
    csr_matrix = sparse_matrix.tocsr()
    
    # Calculate statistics
    row_sums = np.array(csr_matrix.sum(axis=1)).flatten()
    # indptr (index pointer) is array of starting and ending indices in row
    # so we shows each elemnt pointing to start of rows data in mem amd diff gives the difference between each value, 
    # so we get the number of non zero entries 
    row_nonzeros = np.diff(csr_matrix.indptr)  # Number of non-zeros per row
    
    # Calculate means, handling empty rows
    # below gives zeros matrix in same shape as row_sums
    row_means = np.zeros_like(row_sums, dtype=float)
    # makes sure we have no zero ratings which could mess the average up 
    mask = row_nonzeros > 0
    row_means[mask] = row_sums[mask] / row_nonzeros[mask]
    
    return {
        'activity_sum': row_sums,      # Total activity per user
        'interaction_count': row_nonzeros,  # Number of interactions per user
        'average_value': row_means     # Average value per user
    }
```

```python
from scipy.sparse import coo_matrix

# NOTE: User and Movie IDs must be zero-indexed and sequential for the sparse matrix!

# 1. Map IDs to Zero-Indexed Indices
# This creates sequential indices starting from 0, which is mandatory for sparse matrices.
user_map = {id: i for i, id in enumerate(ratings_df['userId'].unique())}
movie_map = {id: i for i, id in enumerate(ratings_df['movieId'].unique())}

ratings_df['user_index'] = ratings_df['userId'].map(user_map)
ratings_df['movie_index'] = ratings_df['movieId'].map(movie_map)

# 2. Build the Sparse Matrix Directly
# The data, rows, and columns correspond to the rating, user_index, and movie_index. this is coo matrix way for creating spares matrices 
sparse_matrix = coo_matrix(
    (ratings_df['rating'], (ratings_df['user_index'], ratings_df['movie_index'])),
    shape=(len(user_map), len(movie_map))
)

# 3. Convert to CSR for Analysis (as required by your function)
# csr_matrix_for_analysis = sparse_matrix.tocsr() 

# 4. Now, run your analysis function:
user_stats = analyze_row_patterns(sparse_matrix)
```

```python
# from above gets the sparse matrix
import numpy as np
from sklearn.decomposition import TruncatedSVD
# truncatedsvd is algorithm for dimension reduction and matrix factorisation for large sparse data
# n_components is a hyperparameter which defines number latent factors i want to extract, 
#   this means each user with vector of 100 latent factors and each movie with vector of 100 latent factors and random state choosen
svd = TruncatedSVD(n_components=100, random_state=42)
# the fit_tranform this performs matric factorization and trains the model on user-item matrix to learn latent factors 
reduced_matrix = svd.fit_transform(sparse_matrix)
# the user matrix is in reduced_matrix and the item matrix is stored in svd object (svd.components_)

# predicited is U * V_T
predicted_ratings = np.dot(reduced_matrix, svd.components_)

idx_to_user_id = {i: id for id, i in user_map.items()}
idx_to_movie_id = {i: id for id, i in movie_map.items()}

# matrix to df
predicted_ratings_df = pd.DataFrame(
    predicted_ratings,
    index=list(idx_to_user_id.values()),   # User IDs as rows
    columns=list(idx_to_movie_id.values()) # Movie IDs as columns
)

predicted_ratings_df.index.name = 'userId'
predicted_ratings_df.columns.name = 'movieId'
```

```python
def recommend_movies_svd(target_user_id, predicted_df, original_ratings_df, movie_metadata_df, top_n=10):
    
    # 1. Get all movies already rated by the target user
    rated_movies = original_ratings_df[original_ratings_df['userId'] == target_user_id]['movieId'].tolist()

    # 2. Get the row of predicted ratings for the target user
    user_predictions = predicted_df.loc[target_user_id].sort_values(ascending=False)

    # 3. Filter out movies the user has already rated
    # gets the user predictions where the rated movies hasn't been watched as the ~ is not and checks in index if rated movie id is not in there
    unseen_predictions = user_predictions[~user_predictions.index.isin(rated_movies)]

    # 4. Get the top N recommended movie IDs
    top_recommendations = unseen_predictions.head(top_n)
    
    # 5. Merge with movie metadata to get titles and genres
    rec_df = movie_metadata_df[
        movie_metadata_df['movieId'].isin(top_recommendations.index)
    ]
    
    # Join with predicted ratings and display
    rec_df = rec_df.set_index('movieId')
    rec_df['Predicted_Rating'] = top_recommendations
    
    # Sort by the predicted rating and return the final list
    return rec_df.sort_values('Predicted_Rating', ascending=False)[['title', 'genres', 'Predicted_Rating']]
```

```python
from fuzzywuzzy import fuzz 
def find_best_match(movielens_title, tmdb_titles_list):
    """
    Finds the best matching TMDb title for a given MovieLens title.
    """
    best_score = 0
    best_match = None
    
    for tmdb_title in tmdb_titles_list:
        # Calculate the Token Set Ratio (robust against missing/extra words)
        score = fuzz.token_set_ratio(movielens_title.lower(), tmdb_title.lower())
        
        if score > best_score:
            best_score = score
            best_match = tmdb_title
            
    # Set a minimum acceptable threshold (e.g., 90)
    if best_score >= 90:
        return best_match, best_score
    else:
        return None, best_score
    
def clean_title(title):
    return title.split(' (')[0].lower()
```

```python

from sklearn.feature_extraction.text import TfidfVectorizer
# tf (term freuqnecy), how often word appears in doc 
# idf (inverse document frequency), how rare word is across all docs 
vectorizer = TfidfVectorizer(
    stop_words='english',  # Automatically remove common English words
    ngram_range=(1, 6),    # Include unigrams (single words) and bigrams (two-word phrases) so tw oword phrases (space adventure) score together 
    min_df=6               # Ignore terms that appear in less than 5 documents, word or phrase must appear at least in 5 different documents
                            # rare words are ignored
)
tfidf_matrix = vectorizer.fit_transform(   #fit gets all vocab that meet criteria and transform convers doc into vector of 
    full_movie_df["overview"].fillna("")  #tf-idf scores so rows = movies | columns = words/bigrams | values = tf-idf weight 
)
print(tfidf_matrix.shape)
tfidf_matrix_32 = tfidf_matrix.astype('float32')
```

```python
from sklearn.metrics.pairwise import linear_kernel
# gets the dot product of the tf-idf matrix with its transpose. So rows is movies, columns is movies and values is similarity score 
cosine_sim = linear_kernel(tfidf_matrix_32)

print("Calculation complete. Shape:", cosine_sim.shape)
```

```python
def genre_check(movie_genre_string, required_gener, min_count=3):
    if pd.isna(movie_genre_string) or pd.isna(required_gener):
        return False
    movie_genre_string= set(movie_genre_string.split('|'))
    return len(required_gener.intersection(movie_genre_string)) >= min_count
```
